```julia
import Pkg
Pkg.add("CSV")
Pkg.add("DataFrames")
```

```julia
using CSV, DataFrames
```

```julia
# Carga el archivo CSV en un DataFrame
df = CSV.read("../input/apartments.csv", DataFrame)
```

```julia
# 1. Crear variable area2 que es el cuadrado de area (0.25 puntos)
df.area2 = df.area .^ 2
```

```julia
# 2. Convertir a dummies (1 para "yes", 0 para "no")
for col in [:hasparkingspace, :hasbalcony, :haselevator, :hassecurity, :hasstorageroom]
    df[!, col] .= Int.(coalesce.(df[!, col], "no") .== "yes")
end
```

```julia

# 3. Crear dummies end_0 a end_9 según último dígito de area
for i in 0:9
    df[!, Symbol("end_", i)] = Int.(floor.(df.area) .% 10 .== i)
end
```

```julia
#LINEAR MODEL ESTIMATION
```

```julia
using Pkg
Pkg.add("GLM")
Pkg.add("StatsModels")
Pkg.add("CategoricalArrays")
```

```julia
using CSV, DataFrames, GLM, StatsModels, CategoricalArrays, Statistics
```

```julia
# Variables predictoras
dropcols = [:price, :id, :end_9]  # end_9 es base
X_vars = setdiff(names(df), dropcols)
```

```julia
first(df[:, X_vars], 5)  # Primeras 5 filas de las variables explicativas
```

```julia
# Esto es crucial para que GLM las trate como variables dummy automáticamente.
categorical_cols = ["month", "type", "rooms", "ownership", "buildingmaterial"]
for col in categorical_cols
    # Asegurarse de que la columna exista
    if col in names(df) # <-- CAMBIAR 'data' por 'df' aquí
        df[!, col] = categorical(df[!, col]) # <-- CAMBIAR 'data' por 'df' aquí
    end
end
println("\nColumnas categóricas convertidas a tipo CategoricalArray.")
println(eltype.(eachcol(df[!, categorical_cols]))) # <-- CAMBIAR 'data' por 'df' aquí
```

```julia
first(df[:, X_vars], 5)  # Primeras 5 filas de las variables explicativas
```

```julia
X_vars = setdiff(X_vars, ["id"])
X_syms = Symbol.(X_vars)
```

```julia
first(df[:, X_vars], 5)  # Primeras 5 filas de las variables explicativas
```

```julia
formula_str = "price ~ " * join(X_vars, " + ")
formula = eval(Meta.parse("@formula(" * formula_str * ")"))
model_ols = lm(formula, df)
```

```julia
println("=== Modelo OLS completo ===")
display(coeftable(model_ols))
```

```julia
# ======= Parte 2: Partialling-out para end_0 =======
X_partial = setdiff(X_vars, [:end_0])
```

```julia
# Crear fórmulas para partialling-out
formula_y_str = "price ~ " * join(X_partial, " + ")
formula_d_str = "end_0 ~ " * join(X_partial, " + ")

form_y = eval(Meta.parse("@formula(" * formula_y_str * ")"))
form_d = eval(Meta.parse("@formula(" * formula_d_str * ")"))
```

```julia
res_y = residuals(lm(form_y, df))
res_d = residuals(lm(form_d, df))
```

```julia
partial_df = DataFrame(e_y = res_y, e_d = res_d)
model_po = lm(@formula(e_y ~ e_d), partial_df)
```

```julia
println("=== Modelo partialling-out ===")
display(coeftable(model_po))
```
