```python
# --- 1) Escribir el archivo LaTeX ---
latex = r"""
%!TEX TS-program = pdflatex
\documentclass[11pt]{article}
\usepackage{amsmath, amssymb, bm}
\usepackage[margin=1in]{geometry}
\title{Frisch--Waugh--Lovell (FWL) Theorem: Detailed Proof}
\date{}
\begin{document}\maketitle

\paragraph{Set-up.}
Let $y\in\mathbb{R}^n$, $X_1\in\mathbb{R}^{n\times k_1}$, $X_2\in\mathbb{R}^{n\times k_2}$ with
$[X_1\;X_2]$ full column rank. The model is
\[
y=X_1\beta_1+X_2\beta_2+u,\qquad \mathbb{E}[u\mid X_1,X_2]=0.
\]
Define the projection and residual-makers on $\mathcal{C}(X_2)$:
\[
P_{X_2}=X_2(X_2'X_2)^{-1}X_2',\qquad M_{X_2}=I-P_{X_2}.
\]
Recall $P_{X_2}=P_{X_2}'=P_{X_2}^2$, $M_{X_2}=M_{X_2}'=M_{X_2}^2$, and $P_{X_2}M_{X_2}=0$.

\paragraph{Goal.}
Show that the OLS coefficient on $X_1$ from the full regression of $y$ on $[X_1\;X_2]$ equals the OLS coefficient from regressing $\tilde y=M_{X_2}y$ on $\tilde X_1=M_{X_2}X_1$:
\[
\hat\beta_1^{\,\text{full}}=\hat\beta_1^{\,\text{FWL}}
=(X_1' M_{X_2}X_1)^{-1}X_1'M_{X_2}y .
\]

\paragraph{Full OLS normal equations.}
OLS minimizes $\|y-X_1\beta_1-X_2\beta_2\|^2$, yielding
\begin{align}
X_1'(y-X_1\hat\beta_1-X_2\hat\beta_2)&=0, \label{ne1}\\
X_2'(y-X_1\hat\beta_1-X_2\hat\beta_2)&=0. \label{ne2}
\end{align}
From \eqref{ne2} we get
\[
X_2'y = X_2'X_1\hat\beta_1 + X_2'X_2\hat\beta_2
\quad\Rightarrow\quad
\hat\beta_2=(X_2'X_2)^{-1}X_2'(y-X_1\hat\beta_1).
\]
Plug into \eqref{ne1}:
\[
X_1'\!\left[y-X_1\hat\beta_1-X_2(X_2'X_2)^{-1}X_2'(y-X_1\hat\beta_1)\right]=0
\;\;\Longleftrightarrow\;\;
X_1' M_{X_2}(y-X_1\hat\beta_1)=0,
\]
hence
\[
(X_1'M_{X_2}X_1)\hat\beta_1=X_1'M_{X_2}y.
\]
Because $[X_1\;X_2]$ has full column rank, $X_1'M_{X_2}X_1$ is nonsingular, so
\[
\boxed{\;\hat\beta_1=(X_1'M_{X_2}X_1)^{-1}X_1'M_{X_2}y\;}.
\]

\paragraph{Two-step residualization interpretation.}
Define residualized variables $\tilde y=M_{X_2}y$ and $\tilde X_1=M_{X_2}X_1$.
OLS of $\tilde y$ on $\tilde X_1$ solves $\tilde X_1'(\tilde y-\tilde X_1\beta_1)=0$, i.e.
\[
(X_1'M_{X_2}X_1)\beta_1=X_1'M_{X_2}y.
\]
Thus its solution coincides with the boxed expression above. Therefore
\[
\hat\beta_1^{\,\text{FWL}}=\hat\beta_1^{\,\text{full}}.
\]

\paragraph{Geometric view (optional).}
Decompose $y=P_{X_2}y + M_{X_2}y$. The component $P_{X_2}y$ lies in $\mathcal{C}(X_2)$ and is orthogonal to any vector in $\mathcal{C}(M_{X_2}X_1)$. Hence only $M_{X_2}y$ carries information about $\beta_1$ once $X_2$ is controlled for, which is why regressing $\tilde y$ on $\tilde X_1$ suffices.

\paragraph{Assumptions for identification and inference.}
Full column rank of $[X_1\;X_2]$ ensures invertibility of $X_1'M_{X_2}X_1$; standard exogeneity and regularity give consistency and usual variance formulas, e.g.
\[
\widehat{\mathrm{Var}}(\hat\beta_1)=\hat\sigma^2\,(X_1'M_{X_2}X_1)^{-1},\quad
\hat\sigma^2=\frac{\|M_{[X_1\,X_2]}y\|^2}{n-k_1-k_2}.
\]
\hfill$\square$

\end{document}
"""

with open("fwl_proof.tex", "w", encoding="utf-8") as f:
    f.write(latex)

print("Archivo escrito: fwl_proof.tex")
```

```python
# --- 2) Verificación numérica rápida de FWL ---
import numpy as np
rng = np.random.default_rng(0)
n, k1, k2 = 500, 2, 3
X1 = rng.standard_normal((n, k1))
X2 = rng.standard_normal((n, k2))
beta1 = np.array([1.0, -2.0])
beta2 = np.array([0.5, -1.0, 0.8])
u = 0.3 * rng.standard_normal(n)
y = X1@beta1 + X2@beta2 + u
```

```python
# Full OLS
X = np.concatenate([X1, X2], axis=1)
bh = np.linalg.lstsq(X, y, rcond=None)[0]
bh1_full = bh[:k1]
```

```python
# FWL
M2 = np.eye(n) - X2 @ np.linalg.inv(X2.T@X2) @ X2.T
X1_t = M2 @ X1
y_t  = M2 @ y
bh1_fwl = np.linalg.lstsq(X1_t, y_t, rcond=None)[0]

print("||b1_full - b1_fwl|| =", np.linalg.norm(bh1_full - bh1_fwl))
```
